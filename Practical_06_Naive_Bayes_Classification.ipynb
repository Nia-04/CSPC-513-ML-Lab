# Practical 6: Naïve Bayes Classification

## Aim

To implement the Naïve Bayes classification algorithm to predict whether to play or not based on weather conditions, namely Outlook and Windy.


## Introduction

Naïve Bayes is a supervised machine learning algorithm used for classification tasks. It is based on Bayes' theorem and calculates the probability of a class based on the given input features.

In this practical, the algorithm is used to predict whether a person should play or not based on two weather conditions: Outlook and Windy.


## Theory

### Bayes' Theorem

Bayes' theorem calculates the probability of an event using prior knowledge and observed evidence.

$$
P(C|X)=\frac{P(X|C)\times P(C)}{P(X)}
$$

Where:

* \(P(C|X)\): Posterior probability of class \(C\) given input \(X\).
* \(P(X|C)\): Likelihood of observing input \(X\) given class \(C\).
* \(P(C)\): Prior probability of class \(C\).
* \(P(X)\): Probability of the input features.

### Naïve Bayes Classifier

The Naïve Bayes classifier assumes that the input features are conditionally independent given the target class.

For this practical, the input features are:

* **Outlook:** Sunny, Overcast, or Rainy.
* **Windy:** Yes or No.

The target variable is **Play**, which can be Yes or No.

The `CategoricalNB` classifier from Scikit-learn is used because the dataset contains categorical features.


```python

import pandas as pd
from sklearn.preprocessing import OrdinalEncoder
from sklearn.naive_bayes import CategoricalNB
```

## Dataset Description

The dataset contains 10 records with three columns:

| Attribute | Description                                  |
| --------- | -------------------------------------------- |
| Outlook   | Weather condition: Sunny, Overcast, or Rainy |
| Windy     | Indicates whether it is windy: Yes or No     |
| Play      | Target class: Yes or No                      |

The dataset is provided in the practical question. The first two attributes are used as input features, and Play is the output variable.


```python

data = {
    'Outlook': [
        'Sunny', 'Sunny', 'Overcast', 'Rainy', 'Rainy',
        'Rainy', 'Overcast', 'Sunny', 'Rainy', 'Overcast'
    ],
    'Windy': [
        'No', 'Yes', 'No', 'No', 'No',
        'Yes', 'Yes', 'No', 'Yes', 'No'
    ],
    'Play': [
        'No', 'No', 'Yes', 'Yes', 'Yes',
        'No', 'Yes', 'Yes', 'Yes', 'Yes'
    ]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)
```

```python

X = df[['Outlook', 'Windy']]
y = df['Play']

encoder = OrdinalEncoder(
    categories=[
        ['Sunny', 'Overcast', 'Rainy'],
        ['No', 'Yes']
    ]
)

X_encoded = encoder.fit_transform(X)

model = CategoricalNB(alpha=1.0)
model.fit(X_encoded, y)

print("Model trained successfully!")
print("Total records:", len(df))

print("\nClass distribution:")
print(y.value_counts())
```

## Prediction for the Given Instance

The instance provided in the question is:

$$
X=(\text{Outlook=Rainy},\ \text{Windy=Yes})
$$

The trained model predicts whether the output class **Play** is Yes or No. The class probabilities are also displayed to show the model's estimated confidence for each class.


```python

test_instance = pd.DataFrame({
    'Outlook': ['Rainy'],
    'Windy': ['Yes']
})

encoded_instance = encoder.transform(test_instance)

prediction = model.predict(encoded_instance)[0]
probabilities = model.predict_proba(encoded_instance)[0]

print("Test Instance: (Rainy, Yes)")
print("Predicted Play:", prediction)

print("\nClass Probabilities:")
for label, probability in zip(model.classes_, probabilities):
    print(f"Play = {label}: {probability:.4f} ({probability * 100:.2f}%)")
```

## Prediction for a New User Input

The program accepts Outlook and Windy values from the user. It validates the entered values, encodes the new instance using the same encoder, and passes it to the trained Naïve Bayes model.

The program displays the predicted Play class and the probabilities for both Yes and No.


```python

print("Enter weather details to predict whether to play.")

outlook = input("Enter Outlook (Sunny/Overcast/Rainy): ").strip().title()
windy = input("Is it Windy? (Yes/No): ").strip().title()

if outlook not in ['Sunny', 'Overcast', 'Rainy'] or windy not in ['Yes', 'No']:
    print("Invalid input! Please enter the allowed values.")

else:
    user_instance = pd.DataFrame({
        'Outlook': [outlook],
        'Windy': [windy]
    })

    encoded_user = encoder.transform(user_instance)
    result = model.predict(encoded_user)[0]
    probabilities = model.predict_proba(encoded_user)[0]

    print("\n--- Prediction Result ---")
    print("Outlook:", outlook)
    print("Windy:", windy)
    print("Predicted Play:", result)

    print("\nClass Probabilities:")
    for label, probability in zip(model.classes_, probabilities):
        print(f"Play = {label}: {probability * 100:.2f}%")
```

## Conclusion

The Naïve Bayes classification algorithm was successfully implemented using Python and Scikit-learn. The model predicts whether to play based on Outlook and Windy. It also accepts new instances from the user and displays the predicted class along with the class probabilities.
