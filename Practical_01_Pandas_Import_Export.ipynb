# Practical 01: Pandas Import and Export

## Aim
To perform data import and export operations using the Pandas library in Python with different file formats such as Excel, CSV, JSON, and TSV.

## Objective

- To understand the use of Pandas for handling external datasets.
- To import data from different file formats into a Pandas DataFrame.
- To export a Pandas DataFrame into different file formats.
- To understand the use of functions such as `read_excel()`, `read_csv()`, `read_json()`, `to_excel()`, `to_csv()`, and `to_json()`.

## Introduction

Pandas is an open-source Python library used for data manipulation and analysis.
It provides powerful data structures such as Series and DataFrame for handling structured data.

A Pandas DataFrame can be imported from and exported to various file formats.
This practical demonstrates the import and export of student data using Excel, CSV, JSON, and TSV formats.

The general workflow is:

**External File → Pandas DataFrame → Processing/Analysis → Exported File**

## 1. Importing the Required Library

```python
# Practical 01
import pandas as pd
```

## 2. Importing Data from Excel

The `read_excel()` function is used to read data from an Excel file and store it in a Pandas DataFrame.

```python
df = pd.read_excel(
    "DATASETS/Mldata.xlsx",
    dtype={"ROLL NO.": "string"} #because python was reading roll numbers as some sort of float data type and changing it into scientific notations
)

df
```

## 3. Exporting Data to Excel

The `to_excel()` function is used to export a Pandas DataFrame into an Excel file.

```python
df.to_excel("EXPORTS/Mldata_export.xlsx", index=False)

print("Data exported successfully to Excel.")
```

## 4. Exporting Data to CSV

The `to_csv()` function is used to export the DataFrame into a CSV (Comma-Separated Values) file.

The parameter `index=False` prevents the Pandas index from being written as an additional column.

```python
df.to_csv("EXPORTS/Mldata.csv", index=False)

print("Data exported successfully to CSV.")
```

### Importing the CSV File

```python
df_csv = pd.read_csv(
    "EXPORTS/Mldata.csv",
    dtype={"ROLL NO.":"string"}
)

df_csv
```

## 5. Exporting Data to JSON

JSON (JavaScript Object Notation) is a commonly used format for storing and exchanging structured data.

The `to_json()` function converts a Pandas DataFrame into JSON format.

The `orient="records"` parameter represents each row as a separate JSON object.

```python
df.to_json("EXPORTS/Mldata.json", orient="records", indent=4)

print("Data exported successfully to JSON.")
```

### Importing the JSON File

```python
df_json = pd.read_json(
    "EXPORTS/Mldata.json",
    dtype={"ROLL NO.": "string"}
)

df_json
```

## 6. Exporting Data to TSV

TSV stands for Tab-Separated Values. It is similar to CSV, but the values are separated using a tab character (`\t`) instead of a comma.

The `sep="\t"` parameter specifies the tab character as the separator.

```python
df.to_csv(
    "EXPORTS/Mldata.tsv",
    sep="\t",
    index=False
)

print("Data exported successfully to TSV.")
```

### Importing the TSV File

```python
df_tsv = pd.read_csv(
    "EXPORTS/Mldata.tsv",
    sep="\t",
    dtype={"ROLL NO.": "string"}
)

df_tsv
```
