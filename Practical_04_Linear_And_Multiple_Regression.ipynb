# Practical 4: Linear Regression and Multiple Linear Regression

## Aim

To implement Linear Regression and Multiple Linear Regression for predicting a student's Data Warehouse marks using the available subject marks.

## Introduction

Regression is a supervised machine learning technique used to predict a continuous numerical value based on one or more input variables.

In this practical, two regression techniques are implemented:

1. **Linear Regression** – uses one input feature to predict the target variable.
2. **Multiple Linear Regression** – uses multiple input features to predict the target variable.

The student marks dataset is used for both models. Data Warehouse marks are selected as the target variable because marks are numerical and continuous.

For Linear Regression, Machine Learning marks are used as the input feature.

For Multiple Linear Regression, Software Engineering, Theory of Computation, Computer Network, and Machine Learning marks are used as input features.

## Linear Regression

Linear Regression is a supervised learning algorithm used to model the relationship between one independent variable and one dependent variable.

The model fits a straight line to the given data and uses this relationship to make predictions.

The general equation of simple linear regression is:

\[
y = b_0 + b_1x
\]

Where:

- \(y\) = predicted output
- \(x\) = input feature
- \(b_0\) = intercept
- \(b_1\) = coefficient or slope

In this practical:

- \(x\) = Machine Learning marks
- \(y\) = Data Warehouse marks

## Multiple Linear Regression

Multiple Linear Regression is an extension of Linear Regression that uses two or more independent variables to predict a single dependent variable.

The general equation is:

**y = b₀ + b₁x₁ + b₂x₂ + ... + bₙxₙ**

Where:

- **y** = predicted output
- **b₀** = intercept
- **b₁, b₂, ..., bₙ** = coefficients
- **x₁, x₂, ..., xₙ** = input features

In this practical:

- **x₁** = Software Engineering marks
- **x₂** = TOC marks
- **x₃** = Computer Network marks
- **x₄** = Machine Learning marks
- **y** = Data Warehouse marks

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
```

```python
import os

if os.path.exists("EXPORTS/Mldata_Preprocessed.xlsx"):
    dataset_path = "EXPORTS/Mldata_Preprocessed.xlsx"
else:
    !git clone https://github.com/Nia-04/CSPC-513-ML-Lab.git
    %cd CSPC-513-ML-Lab
    dataset_path = "EXPORTS/Mldata_Preprocessed.xlsx"

df = pd.read_excel(dataset_path)

df.head()
```

```python
df.shape
```

```python
df.columns
```

## Linear Regression Implementation

For Linear Regression, Machine Learning marks are selected as the independent variable and Data Warehouse marks are selected as the dependent variable.

The complete dataset is used for training the model.

```python
X_linear = df[['MACHINE LEARNING']]
y_linear = df['DATA WAREHOUSE']

X_linear.head()
```

```python
linear_model = LinearRegression()

linear_model.fit(X_linear, y_linear)
```

```python
print("Intercept:", linear_model.intercept_)
print("Coefficient:", linear_model.coef_[0])
```

```python
print(
    f"Equation: Data Warehouse = "
    f"{linear_model.intercept_:.2f} + "
    f"({linear_model.coef_[0]:.2f} × Machine Learning)"
)
```

### User Input for Linear Regression

The trained Linear Regression model is used to predict Data Warehouse marks for a new student.

The user enters the Machine Learning marks, and the model predicts the corresponding Data Warehouse marks.

```python
ml_marks = float(input("Enter Machine Learning marks out of 60: "))
print("You entered:",ml_marks)
```

```python
if ml_marks < 0 or ml_marks > 60:
    raise ValueError("Marks must be between 0 and 60.")
```

```python
new_student_linear = pd.DataFrame(
    [[ml_marks]],
    columns=['MACHINE LEARNING']
)
```

```python
linear_prediction = linear_model.predict(new_student_linear)[0]

print(f"Predicted Data Warehouse marks: {linear_prediction:.2f}")
```

## Multiple Linear Regression Implementation

For Multiple Linear Regression, four subject marks are selected as independent variables and Data Warehouse marks are selected as the dependent variable.

The input features are:

- Software Engineering
- TOC
- Computer Network
- Machine Learning

The complete dataset is used to train the model.

```python
X_multiple = df[
    [
        'SOFTWARE ENGINEER',
        'TOC',
        'COMPUTER NETWORK',
        'MACHINE LEARNING'
    ]
]

y_multiple = df['DATA WAREHOUSE']
```

```python
multiple_model = LinearRegression()

multiple_model.fit(X_multiple, y_multiple)
```

```python
print("Intercept:", multiple_model.intercept_)

print("\nCoefficients:")

for feature, coefficient in zip(
    X_multiple.columns,
    multiple_model.coef_
):
    print(f"{feature}: {coefficient:.4f}")
```

### User Input for Multiple Linear Regression

The user enters marks for Software Engineering, TOC, Computer Network, and Machine Learning.

The trained Multiple Linear Regression model then predicts the student's Data Warehouse marks.

```python
print("Enter marks obtained out of 60")
print()

software_engineer = float(input("Software Engineering: "))
toc = float(input("Theory of Computation (TOC): "))
computer_network = float(input("Computer Network: "))
machine_learning = float(input("Machine Learning: "))
print("User Inputs:")
print("Software Engineering: ",software_engineer)
print("Theory of Computation (TOC): ",toc)
print("Computer Network: ",computer_network)
print("Machine Learning: ",machine_learning)
```

```python
marks = [
    software_engineer,
    toc,
    computer_network,
    machine_learning
]

if any(mark < 0 or mark > 60 for mark in marks):
    raise ValueError("Marks must be between 0 and 60.")
```

```python
new_student_multiple = pd.DataFrame(
    [[
        software_engineer,
        toc,
        computer_network,
        machine_learning
    ]],
    columns=X_multiple.columns
)
```

```python
multiple_prediction = multiple_model.predict(
    new_student_multiple
)[0]

print(
    f"Predicted Data Warehouse marks: "
    f"{multiple_prediction:.2f}"
)
```

## Prediction Comparison

The predictions obtained from Linear Regression and Multiple Linear Regression are displayed for comparison.

Linear Regression uses only Machine Learning marks, while Multiple Linear Regression uses four subject marks.

```python
print("Prediction Results")
print("----------------------------")
print(
    f"Linear Regression: "
    f"{linear_prediction:.2f}"
)
print(
    f"Multiple Linear Regression: "
    f"{multiple_prediction:.2f}"
)
```
